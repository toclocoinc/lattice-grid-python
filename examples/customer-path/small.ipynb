```python
# A 200-row DataFrame, small mode: the rows travel to the grid.
import pandas as pd
from lattice_grid_jupyter import LatticeGridWidget

df = pd.DataFrame({
    "id": range(200),
    "region": ["north", "south", "east", "west"] * 50,
    "score": [(i * 37) % 200 for i in range(200)],
    "trend": [[i % 5, (i + 1) % 5, (i + 2) % 5] for i in range(200)],
})
w = LatticeGridWidget(df, offline=True)
print("MODE=%s" % ("windowed" if w.windowed else "client"))
```

```python
w
```

```python
# Sort a column and filter another, the way the state dict documents it.
w.state = {"version": 2,
           "sort": [{"col": "score", "dir": "desc"}],
           "filters": {"col": "region", "op": "eq", "value": "north"}}
```

```python
# A later cell reads the filtered, sorted frame.
view = w.view
want = df[df["region"] == "north"].sort_values("score", ascending=False)
assert list(view["id"]) == list(want["id"]), (list(view["id"])[:5], list(want["id"])[:5])
assert list(view["score"]) == sorted(view["score"], reverse=True)
assert set(view["region"]) == {"north"}
assert all(isinstance(t, list) for t in view["trend"])
print("SMALL_ROWS=%d" % len(view))
print("SMALL_TOP=%s" % list(view["score"].head(3)))
print("SMALL_FIRST_ID=%d" % view["id"].iloc[0])
```
